```python
# data larni clean ing qilish da(tozalash da ishlatiladi)
data_with_missing ={
    "cars": ["BMW", None, "Ford", "Tesla", None, "Lambo", "GM", None],
    "counts": [None, 7, 2, 4, 9, None, 2, None]
}

df_missing =  pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull() # data larni clean ing qilish da(tozalash da ishlatiladi)
# boolen da True(data yoq bolsa)/False(data bor bolsa) da chiqarib beradi
# df_missing.isnull().sum() ni ham qoshadigan bolsak, data yoq row larni soni ni chiqaradi
```

```python
df_missing.dropna() # malumotlarni ochirish uchun
# data si yoq qatorni ochirib tashlaydi, data si toliqlarni chiqarib beradi
# df_missing.dropna(how="all") toliq missing value bor qatorni faqat ochiradi
```

```python
df_missing.dropna(axis=1) #malumoti yoq hamma column larni tashlab yuboradi,
#faqat column boyicha tekshiradi
# df_missing.dropna(axis=1) qaysi clumn da 1 ta data bolmasa ham ochirib tashlaydi
# df_missing.dropna(subset=['counts']) faqat counts column ni tekshirib beradi
```

```python
df_missing.fillna(7) # value si yoq qatorlarga malumot qoyib qoyadi
# missing value lar ga argument da berilgan qiymatni  qoyib chiqadi
# df_missing.fillna("unknown") string bolishi ham m-
# df_missing.fillna("cars": "unknown", "counts": 0)  column lardan kelib chiqib value kiritishimiz mumkun
```
