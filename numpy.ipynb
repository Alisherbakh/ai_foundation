```python
import numpy as np

arr = np.array([1,3,5,7])
print('numpy_array:', arr)

arr.shape
```

```python
arr2 = np.array([[1,3,5,7], [4,6,8,9]])
print('numpy_array2:', arr2)

arr2.shape # shaklini korish uchun, 2 qator, 4 ustun
```

```python
#dtype() arraydagi elementlarni type ni aniqlab beradi
a = np.array([1,2,3])
print(a.dtype)
```

```python
# reshape() yaratilgan arrayni shape ni ozgartirib olish uchun, projectda malum data type kerak bolsa,
# image data larni shaklini ozgartirib olish kk bolishi mumkun
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print("shape_array:", b)
```

```python
# astype() data type ni ozgartirib beradi, matematik amallarda va bir xil turdagi data type kerak bolsa
a = np.array([1,2,3])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeros(), np.ones() => dummy data yani test data lar hosil qilishda ishlatiladi
z = np.zeros((2,3))
o = np.ones((2,3))

print("zeros:", z)
print("ones:", o)
```

```python
#  shu dataset da arrayda nechta dimension yani olchov borligini korsatib beradi 1D or 2D
a = np.array([[1,2], [3,4]])
print("indim:", a.ndim)
```

```python
# arange() ketma ket sonlar yaratib beriladi
a = np.arange(1,7)
print("arange:", a)
```

```python
# sum(), mean(), max(), min() => statistikalarda ishlatiladi, malumotlarni ozimizga moslab olishda ishlatiladi
# +  , ortachasini, eng kattaa, eng kichik

a = np.array([1,2,3,4])

print("sum:", a.sum())
print("mean:", a.mean())
print("max:", a.max())
print("min:", a.min())
```

```python
# flatten 2D yoki 3D arrayni 1D ga ozzgartitb beradi, machine learningda

a = np.array([[1,2], [3,4]])
print("flatten:",  a.flatten())
```

```python
# T (Transpose) array shaklini ozgartirishda, malumotlarni qayta tashkil qillishda
a = np.array([[1,2],[3,4]])
print("transpose:", a.T)
```

```python
# dot() @ matrix ni kopaytirishda ishlatiladi
a = np.array([[1,2],[3,4]])
b = np.array([[2,0], [1,2]])
print("dot/multiplication:", a.dot(b))
```

```python
# random.rand() random raqamlarni yaratib beradi 2 qator 3 ustun
a = np.random.rand(2,3) # => 0.0 and 1.0 / 0.0 included / 1.0 excluded
print("random rand:", a)
```
